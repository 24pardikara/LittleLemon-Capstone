# Little Lemon - Python client

Connects to **LittleLemonDB** with MySQL Connector/Python and demonstrates the five
stored procedures: `GetMaxQuantity`, `ManageBooking`, `AddBooking`, `UpdateBooking`,
`CancelBooking`.

Run `LittleLemonDB.sql` and `load_data.py` first so the database exists and is populated.

```python
import os
import mysql.connector
from mysql.connector import Error

dbconfig = {
    "host": "localhost",
    "user": "root",
    "password": os.getenv("MYSQL_PASSWORD", "your_password"),  # set MYSQL_PASSWORD or edit here
    "database": "LittleLemonDB",
}

connection = mysql.connector.connect(**dbconfig)
cursor = connection.cursor()
print("Connected to LittleLemonDB.")

cursor.execute("SHOW TABLES;")
print("Tables:", [t[0] for t in cursor.fetchall()])
```

## 1. GetMaxQuantity()
Displays the maximum quantity ordered in a single order.

```python
cursor.callproc("GetMaxQuantity")
for result in cursor.stored_results():
    print("Max quantity in order:", result.fetchone()[0])
```

## 2. ManageBooking(date, table)
Checks whether a table is free or already booked on a given date.

```python
cursor.callproc("ManageBooking", ["2022-10-10", 5])
for result in cursor.stored_results():
    print(result.fetchone()[0])

cursor.callproc("ManageBooking", ["2022-10-10", 99])
for result in cursor.stored_results():
    print(result.fetchone()[0])
```

## 3. AddBooking(id, customer, date, table)
Adds a new table booking.

```python
# pick any existing customer id for the foreign key
cursor.execute("SELECT CustomerID FROM Customers LIMIT 1;")
cust = cursor.fetchone()[0]

cursor.callproc("AddBooking", [10, cust, "2022-12-25", 8])
for result in cursor.stored_results():
    print(result.fetchone()[0])
connection.commit()
```

## 4. UpdateBooking(id, date)
Updates the date of an existing booking.

```python
cursor.callproc("UpdateBooking", [10, "2022-12-26"])
for result in cursor.stored_results():
    print(result.fetchone()[0])
connection.commit()
```

## 5. CancelBooking(id)
Removes a booking.

```python
cursor.callproc("CancelBooking", [10])
for result in cursor.stored_results():
    print(result.fetchone()[0])
connection.commit()

# close up
cursor.close()
connection.close()
print("Connection closed.")
```
